```python
import torch
my_torch = torch.arange(10)
my_torch
```

```python
# Reshape and View
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
# Reshaping if we don't know the number of items
# using -1
my_torch2 = torch.arange(10)
print(f"my_torch2 : \n{my_torch2}")

my_torch2 = my_torch2.reshape(5,-1)
print(f"\nAfter reshaping with (5,-1)\n{my_torch2}")

my_torch2 = my_torch2.reshape(-1,5)
print(f"\nAfter reshaping with (-1,5)\n{my_torch2}")
```

```python
# Using View
my_torch3 = torch.arange(10)
print(f"my_torch3 = \n{my_torch3}")

my_torch3 = my_torch3.view(2,5)
print(f"\nAfter using view = \n{my_torch3}")
```
